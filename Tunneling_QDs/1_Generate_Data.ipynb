# Control Panel 1: Generate Data

This notebook generates double quantum dot (DQD) stability diagram datasets with pink noise.

## Usage
1. Configure parameters below
2. Run all cells to generate and save dataset

```python
# Import required modules
import sys
from pathlib import Path

# Add src to path
sys.path.append('src')

from src.config import default_params_5
from src.generator import generate_and_save_dqd_dataset
import numpy as np
import matplotlib.pyplot as plt
```

## Configure Parameters

Modify the parameters below to customize your dataset generation.

```python
# Load default parameters
params = default_params_5.copy()

# Customize parameters as needed
params['num_samples'] = 1000  # Number of diagrams to generate
params['resolution'] = 128    # Grid resolution (128x128)
params['noise_level'] = 0.01  # Pink noise amplitude

# Display current parameters
print("Current parameters:")
for key, value in params.items():
    print(f"  {key}: {value}")
```

## Generate Dataset

Run this cell to generate the dataset. This may take several minutes depending on `num_samples`.

```python
# Generate and save dataset
output_filename = 'dqd_dataset.h5'
num_samples = params['num_samples']

print(f"Generating {num_samples} samples...")
dataset_path = generate_and_save_dqd_dataset(
    num_samples=num_samples,
    filename=output_filename,
    params=params
)

print(f"\nDataset successfully saved to: {dataset_path}")
```

## Visualize Sample Data

Preview some generated stability diagrams to verify data quality.

```python
# Load and visualize a few samples
import h5py

with h5py.File(dataset_path, 'r') as f:
    # Load first 4 samples
    samples = f['diagrams'][:4]
    
    # Plot
    fig, axes = plt.subplots(2, 2, figsize=(10, 10))
    axes = axes.flatten()
    
    for i, (ax, sample) in enumerate(zip(axes, samples)):
        im = ax.imshow(sample, cmap='hot', aspect='auto')
        ax.set_title(f'Sample {i+1}')
        ax.set_xlabel('Gate Voltage 1 (mV)')
        ax.set_ylabel('Gate Voltage 2 (mV)')
        plt.colorbar(im, ax=ax)
    
    plt.tight_layout()
    plt.show()

print(f"\nDataset shape: {samples.shape}")
print(f"Value range: [{samples.min():.3f}, {samples.max():.3f}]")
```
